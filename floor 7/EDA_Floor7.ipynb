## Understanding the Floor 7 columns

The dataset uses a consistent naming pattern:

`z<zone>_<equipment or sensor>(unit)`

For example, `z1_AC1(kW)` means AC unit 1 in zone 1, measured in kilowatts. The zone prefix tells us where the measurement belongs, while the equipment or sensor name tells us what is being measured.


`Date` is the timestamp. `year` is an analysis label added when the 2018 and 2019 files were combined.

```python
import re

measurement_columns = [
    column for column in floor7.columns
    if column not in ['Date', 'year']
]

column_records = []
for column in measurement_columns:
    match = re.match(r'^(z\d+)_(.+)\((.+)\)$', column)
    if not match:
        continue

    zone, measurement, unit = match.groups()

    if measurement.startswith('AC'):
        measurement_group = 'Electrical Power'
        use_case = 'AC'
    elif measurement == 'Light':
        measurement_group = 'Electrical Power'
        use_case = 'Lighting'
    elif measurement == 'Plug':
        measurement_group = 'Electrical Power'
        use_case = 'Plug Load'
    elif measurement == 'S1' and unit == 'degC':
        measurement_group = 'Temperature'
        use_case = 'Temperature'
    elif measurement == 'S1' and unit == 'RH%':
        measurement_group = 'Humidity'
        use_case = 'Humidity'
    elif measurement == 'S1' and unit == 'lux':
        measurement_group = 'Illuminance'
        use_case = 'Illuminance'
    else:
        continue

    column_records.append({
        'column': column,
        'zone': zone,
        'measurement': measurement_group,
        'use_case': use_case,
        'unit': unit
    })

column_dictionary = pd.DataFrame(
    column_records,
    columns=['column', 'zone', 'measurement', 'use_case', 'unit']
)

print('Column dictionary:')
display(column_dictionary)
```

The columns will be grouped in two ways:

- **By zone:** all columns belonging to the same zone.
- **By use case:** in each zone, all columns belonging to the same type of use case.

```python
print('\nColumns grouped by zone and use_case:')
for zone, zone_group in column_dictionary.groupby('zone', sort=True):
    print(f'\n{zone}:')
    for use_case, use_case_group in zone_group.groupby('use_case', sort=True):
        print(f"  {use_case}: {', '.join(use_case_group['column'])}")
```

## Missing-value analysis by columns

All columns in descending order of percentage of missing data.

```python
total_rows = len(floor7)
measurement_data = floor7.drop(columns=['Date', 'year'])

missing_summary = pd.DataFrame({
    'rows_with_data': measurement_data.notna().sum(),
    'rows_missing': measurement_data.isna().sum()
})
missing_summary['missing_percentage'] = (
    missing_summary['rows_missing'] / total_rows * 100
).round(2)

print(f'Total rows analysed: {total_rows:,}')
print('Missing-value summary by column:')
display(missing_summary)
```

```python
import matplotlib.pyplot as plt

plot_data = missing_summary.sort_values(
    'missing_percentage',
    ascending=True
)

plt.figure(figsize=(12, 8))
plt.barh(
    plot_data.index,
    plot_data['missing_percentage'],
    color='steelblue'
)
plt.xlabel('Missing data (%)')
plt.ylabel('Column')
plt.title('Missing Data Percentage by Column')
plt.xlim(0, 100)
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

## Key findings

- **Environmental measurements have the most missing data:** temperature, humidity, and illuminance columns have approximately 33% to 36% missing values.
- **Zone 5 has the highest missingness:** its environmental columns have about 35.64% missing values, the highest among the measurement columns.
- **Electrical-power data is more complete:** AC, lighting, and plug-load columns have between 0.15% and 3.83% missing values.

## Monthly missing-value analysis

TOP 10 months with the highest percentage of missing data.

```python
month_period = floor7['Date'].dt.to_period('M')

monthly_missing_count = (
    measurement_data.isna()
    .groupby(month_period)
    .sum()
    .sum(axis=1)
)

monthly_total_values = (
    measurement_data.groupby(month_period).size()
    * measurement_data.shape[1]
)

monthly_missing_percentage = (
    monthly_missing_count / monthly_total_values * 100
).sort_values(ascending=False)

top_missing_months = monthly_missing_percentage.head(10).sort_values()

plt.figure(figsize=(12, 6))
plt.barh(
    top_missing_months.index.astype(str),
    top_missing_months.values,
    color='darkorange'
)
plt.xlabel('Missing data (%)')
plt.ylabel('Month')
plt.title('Top 10 Months by Missing Data Percentage')
plt.xlim(0, 100)
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

## Key finding

- **The seven highest-missing months form one consecutive period from September 2018 to March 2019, with February 2019 having the highest missing-data percentage at approximately 48.73%.**

## Missing-value heatmap by column and month

Each heatmap cell shows the percentage of missing values for one measurement column during one month.

```python
monthly_missing_by_column = (
    measurement_data.isna()
    .assign(month=floor7['Date'].dt.to_period('M'))
    .groupby('month')
    .mean()
    .mul(100)
    .T
)

monthly_missing_by_column.columns = [
    month.to_timestamp().strftime('%b %Y')
    for month in monthly_missing_by_column.columns
]

plt.figure(figsize=(18, 10))
plt.imshow(
    monthly_missing_by_column,
    aspect='auto',
    cmap='YlOrRd',
    vmin=0,
    vmax=100
)

plt.colorbar(label='Missing data (%)')
plt.xticks(
    range(len(monthly_missing_by_column.columns)),
    monthly_missing_by_column.columns,
    rotation=45,
    ha='right'
)
plt.yticks(
    range(len(monthly_missing_by_column.index)),
    monthly_missing_by_column.index
)
plt.xlabel('Month (MMM YYYY)')
plt.ylabel('Measurement column')
plt.title('Missing Data Percentage by Column and Month')
plt.tight_layout()
plt.show()
```

## Key findings

- **Missingness is concentrated in environmental measurements:** temperature, humidity, and illuminance columns show the darkest and most persistent gaps in the heatmap.
- **The main missing-data period is September 2018 to March 2019:** this pattern affects multiple environmental columns at the same time.
- **Electrical-power measurements remain comparatively complete:** the power columns show mostly low missing-data percentages across the months.

## Comparing maximum values between 2018 and 2019

For each measurement column, we compare the maximum recorded value in 2018 with the maximum recorded value in 2019. The percentage change is calculated relative to the 2018 maximum.

```python
max_2018 = floor7_2018[measurement_columns].max()
max_2019 = floor7_2019[measurement_columns].max()

max_comparison = pd.DataFrame({
    'maximum_2018': max_2018,
    'maximum_2019': max_2019
})

max_comparison['difference'] = (
    max_comparison['maximum_2019'] - max_comparison['maximum_2018']
)
max_comparison['percentage_change'] = (
    max_comparison['difference'] / max_comparison['maximum_2018'].abs() * 100
).replace([float('inf'), -float('inf')], pd.NA).round(2)
max_comparison['absolute_percentage_change'] = (
    max_comparison['percentage_change'].abs()
)

max_comparison = max_comparison.sort_values(
    'absolute_percentage_change',
    ascending=False
)

print('Maximum-value comparison, ranked by absolute percentage change:')
display(max_comparison)

plot_comparison = max_comparison.dropna(subset=['percentage_change']).head(10)
plot_comparison = plot_comparison.sort_values('percentage_change')

plt.figure(figsize=(12, 6))
plt.barh(
    plot_comparison.index,
    plot_comparison['percentage_change'],
    color=['indianred' if value > 0 else 'steelblue'
           for value in plot_comparison['percentage_change']]
)
plt.axvline(0, color='black', linewidth=0.8)
plt.xlabel('Percentage change from 2018 maximum to 2019 maximum (%)')
plt.ylabel('Measurement column')
plt.title('Largest Year-over-Year Changes in Maximum Values')
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

## Key findings

- **The largest increase was in `z3_Plug(kW)`:** its maximum rose from `0.36 kW` in 2018 to `1.12 kW` in 2019, a **211.11% increase**.
- **Lighting maxima also increased substantially:** `z5_Light(kW)`, `z2_Light(kW)`, and `z4_Light(kW)` increased by **129.25%**, **117.57%**, and **103.12%**, respectively.
- **The largest decreases were `z1_AC2(kW)` and `z4_S1(lux)`:** their maxima decreased by **56.97%** and **26.26%**. These maximum-value differences identify potential spikes for further investigation, but do not by themselves confirm faults.

## Comparing minimum values between 2018 and 2019

For each measurement column, we compare the minimum recorded value in 2018 with the minimum recorded value in 2019. Percentage changes are calculated relative to the 2018 minimum; zero 2018 minimums are excluded from percentage calculations.

```python
min_2018 = floor7_2018[measurement_columns].min()
min_2019 = floor7_2019[measurement_columns].min()

min_comparison = pd.DataFrame({
    'minimum_2018': min_2018,
    'minimum_2019': min_2019
})

min_comparison['difference'] = (
    min_comparison['minimum_2019'] - min_comparison['minimum_2018']
)
min_comparison['percentage_change'] = (
    min_comparison['difference'] / min_comparison['minimum_2018'].abs() * 100
).replace([float('inf'), -float('inf')], pd.NA).round(2)
min_comparison['absolute_percentage_change'] = (
    min_comparison['percentage_change'].abs()
)

min_comparison = min_comparison.sort_values(
    'absolute_percentage_change',
    ascending=False
)

print('Minimum-value comparison, ranked by absolute percentage change:')
display(min_comparison)

plot_min_comparison = min_comparison.dropna(
    subset=['percentage_change']
).head(10)
plot_min_comparison = plot_min_comparison.sort_values('percentage_change')

plt.figure(figsize=(12, 6))
plt.barh(
    plot_min_comparison.index,
    plot_min_comparison['percentage_change'],
    color=['indianred' if value > 0 else 'steelblue'
           for value in plot_min_comparison['percentage_change']]
)
plt.axvline(0, color='black', linewidth=0.8)
plt.xlabel('Percentage change from 2018 minimum to 2019 minimum (%)')
plt.ylabel('Measurement column')
plt.title('Largest Year-over-Year Changes in Minimum Values')
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

## Key findings

- **Minimum environmental readings were lower in 2019:** all comparable temperature and humidity columns decreased from their 2018 minimums.
- **The largest decrease was `z2_S1(RH%)`:** its minimum fell from `52.52%` to `39.79%`, a **24.24% decrease**.
- **Power and illuminance minimums were mostly `0` in both years:** percentage differences are therefore undefined for these columns and should not be interpreted as changes.

## Electrical-power load profile

We group the electrical measurements into AC, Lighting, and Plug Load, then calculate the average hourly power across all dates to create a typical daily load profile.

```python
power_groups = {
    'AC': [column for column in measurement_columns if '_AC' in column],
    'Lighting': [column for column in measurement_columns if '_Light' in column],
    'Plug Load': [column for column in measurement_columns if '_Plug' in column]
}

power_by_use_case = pd.DataFrame(index=floor7.index)
for use_case, columns in power_groups.items():
    power_by_use_case[use_case] = floor7[columns].sum(axis=1, min_count=1)

power_by_use_case['hour'] = floor7['Date'].dt.hour
average_daily_power = power_by_use_case.groupby('hour').mean()

plt.figure(figsize=(12, 6))
for use_case in power_groups:
    plt.plot(
        average_daily_power.index,
        average_daily_power[use_case],
        marker='o',
        linewidth=2,
        label=use_case
    )

plt.xticks(range(24))
plt.xlabel('Hour of day')
plt.ylabel('Average electrical power (kW)')
plt.title('Typical Daily Electrical-Power Load Profile by Use Case')
plt.grid(alpha=0.3)
plt.legend(title='Use case')
plt.tight_layout()
plt.show()
```

## Key findings

- **AC is the dominant electrical load:** it rises sharply around 08:00 and 13:00, reaching approximately 67–69 kW.
- **Lighting follows a daytime pattern:** it increases from around 06:00, stays near 10 kW during working hours, and declines after 17:00.
- **Plug-load power is comparatively low and stable:** it remains around 0–1 kW across most hours.

## Zone-level electrical load profiles

Each small chart shows the typical daily electrical load profile for one zone, split into AC, Lighting, and Plug Load.

```python
zone_use_case_columns = {
    f'Zone {zone}': {
        'AC': [
            column for column in measurement_columns
            if column.startswith(f'z{zone}_AC')
        ],
        'Lighting': [
            column for column in measurement_columns
            if column == f'z{zone}_Light(kW)'
        ],
        'Plug Load': [
            column for column in measurement_columns
            if column == f'z{zone}_Plug(kW)'
        ]
    }
    for zone in range(1, 6)
}

zone_average_profiles = {}
for zone, use_case_columns in zone_use_case_columns.items():
    zone_data = pd.DataFrame(index=floor7.index)
    for use_case, columns in use_case_columns.items():
        if columns:
            zone_data[use_case] = floor7[columns].sum(axis=1, min_count=1)
        else:
            zone_data[use_case] = float('nan')

    zone_data['hour'] = floor7['Date'].dt.hour
    zone_average_profiles[zone] = zone_data.groupby('hour').mean()

fig, axes = plt.subplots(2, 3, figsize=(10, 5), sharex=True)
axes = axes.ravel()

for index, (zone, average_profile) in enumerate(zone_average_profiles.items()):
    axis = axes[index]
    for use_case in ['AC', 'Lighting', 'Plug Load']:
        axis.plot(
            average_profile.index,
            average_profile[use_case],
            linewidth=1.3,
            label=use_case
        )

    axis.set_title(zone, fontsize=10)
    axis.set_xticks(range(0, 24, 6))
    axis.grid(alpha=0.3)
    axis.tick_params(labelsize=8)

axes[-1].set_visible(False)
fig.supxlabel('Hour of day')
fig.supylabel('Average power (kW)')
fig.suptitle('Typical Daily Electrical Load Profile by Zone and Use Case', fontsize=13)
axes[0].legend(fontsize=7)
plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
```

## End-use Pareto chart by zone

This horizontal stacked bar chart compares the average electrical-power contribution of AC, Lighting, and Plug Load within each zone. Zones are ordered by total average power.

```python
zone_end_use_average = pd.DataFrame(index=zone_average_profiles.keys())
for zone, average_profile in zone_average_profiles.items():
    zone_end_use_average.loc[zone, 'AC'] = average_profile['AC'].mean()
    zone_end_use_average.loc[zone, 'Lighting'] = average_profile['Lighting'].mean()
    zone_end_use_average.loc[zone, 'Plug Load'] = average_profile['Plug Load'].mean()

zone_end_use_average = zone_end_use_average.fillna(0)
zone_end_use_average['total'] = zone_end_use_average.sum(axis=1)
zone_end_use_average = zone_end_use_average.sort_values('total')

end_use_columns = ['AC', 'Lighting', 'Plug Load']
colors = {
    'AC': '#243b6b',
    'Lighting': '#c9972b',
    'Plug Load': '#8f9aae'
}

fig, axis = plt.subplots(figsize=(10, 4.5))
left = pd.Series(0.0, index=zone_end_use_average.index)

for end_use in end_use_columns:
    values = zone_end_use_average[end_use]
    axis.barh(
        zone_end_use_average.index,
        values,
        left=left,
        label=end_use,
        color=colors[end_use]
    )

    for zone, value in values.items():
        total = zone_end_use_average.loc[zone, 'total']
        if total >= 1 and value / total >= 0.05:
            axis.text(
                left[zone] + value / 2,
                zone,
                f'{value / total:.0%}',
                ha='center',
                va='center',
                color='white',
                fontsize=8
            )

    left += values

axis.set_xlabel('Average electrical power (kW)')
axis.set_ylabel('Zone')
axis.set_title('End-use Pareto Profile by Zone')
axis.grid(axis='x', linestyle='--', alpha=0.35)
axis.legend(title='End use', loc='lower right')
plt.tight_layout()
plt.show()
```

## Key findings

- **Zone 1 has the highest electrical usage:** its AC load peaks at approximately `25 kW`, and AC accounts for **89%** of its average electrical power.
- **Zones 2 and 4 show similar daily profiles:** their dominant AC loads peak at approximately `19–21 kW`, while AC contributes **83%** and **75%** of their average power, respectively.
- **Zone 5 has a lower AC load and a more balanced end-use mix:** AC contributes **56%** and Lighting contributes **42%**. Zone 3 has the lowest total usage and no AC measurements.

## Zone 3 focused analysis

Zone 3 is different from the other zones because it has Lighting and Plug Load measurements only. It has no AC or indoor-environment sensors, so this section focuses on its electrical load profile and monthly variation.

```python
zone3_columns = ['z3_Light(kW)', 'z3_Plug(kW)']
zone3_data = floor7[['Date'] + zone3_columns].copy()
zone3_data['hour'] = zone3_data['Date'].dt.hour
zone3_data['month'] = zone3_data['Date'].dt.to_period('M')

zone3_hourly_profile = zone3_data.groupby('hour')[zone3_columns].mean()
zone3_monthly_profile = zone3_data.groupby('month')[zone3_columns].mean()
zone3_summary = pd.DataFrame({
    'average_kW': zone3_data[zone3_columns].mean(),
    'maximum_kW': zone3_data[zone3_columns].max(),
    'missing_percentage': zone3_data[zone3_columns].isna().mean().mul(100).round(2)
})

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

for column in zone3_columns:
    axes[0].plot(
        zone3_hourly_profile.index,
        zone3_hourly_profile[column],
        marker='o',
        linewidth=1.8,
        label=column.replace('z3_', '').replace('(kW)', '')
    )

axes[0].set_title('Zone 3 Daily Load Profile')
axes[0].set_xlabel('Hour of day')
axes[0].set_ylabel('Average power (kW)')
axes[0].set_xticks(range(0, 24, 3))
axes[0].grid(alpha=0.3)
axes[0].legend(title='End use')

monthly_labels = [month.to_timestamp().strftime('%b %Y') for month in zone3_monthly_profile.index]
for column in zone3_columns:
    axes[1].plot(
        monthly_labels,
        zone3_monthly_profile[column],
        marker='o',
        linewidth=1.8,
        label=column.replace('z3_', '').replace('(kW)', '')
    )

axes[1].set_title('Zone 3 Monthly Average Power')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Average power (kW)')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(alpha=0.3)
axes[1].legend(title='End use')

fig.suptitle('Zone 3 Electrical Load Profile', fontsize=13)
plt.tight_layout()
plt.show()

display(zone3_summary)
```

## Key findings

- **Zone 3 has very low electrical demand:** the maximum values are `1.12 kW` for Plug Load and `0.89 kW` for Lighting, far below the AC-equipped zones.
- **The monthly Plug Load profile shows a clear drop from January to March 2019:** it falls from approximately `0.21 kW` to `0.06 kW` before rising again in April 2019.

## Indoor conditions across zones

This quick comparison uses monthly mean Temperature, Relative Humidity, and Illuminance for the zones with available indoor-environment sensors: Zones 1, 2, 4, and 5. Zone 3 has no indoor-condition measurements.

```python
environment_groups = {
    'Temperature': [
        column for column in measurement_columns
        if column.endswith('_S1(degC)')
    ],
    'Relative Humidity': [
        column for column in measurement_columns
        if column.endswith('_S1(RH%)')
    ],
    'Illuminance': [
        column for column in measurement_columns
        if column.endswith('_S1(lux)')
    ]
}

environment_summary_rows = []
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

for axis, (measurement, columns) in zip(axes, environment_groups.items()):
    zone_names = [column.split('_')[0].replace('z', 'Zone ') for column in columns]
    environment_data = floor7[['Date'] + columns].copy()
    environment_data['month'] = environment_data['Date'].dt.to_period('M')
    monthly_means = environment_data.groupby('month')[columns].mean()
    month_labels = [month.to_timestamp().strftime('%b %Y') for month in monthly_means.index]

    for column, zone_name in zip(columns, zone_names):
        axis.plot(
            month_labels,
            monthly_means[column],
            marker='o',
            linewidth=1.5,
            label=zone_name
        )

        environment_summary_rows.append({
            'measurement': measurement,
            'zone': zone_name,
            'average': floor7[column].mean(),
            'minimum': floor7[column].min(),
            'maximum': floor7[column].max(),
            'missing_percentage': floor7[column].isna().mean() * 100
        })

    axis.set_title(measurement)
    axis.set_xlabel('Month')
    axis.tick_params(axis='x', rotation=45, labelsize=8)
    axis.grid(alpha=0.3)
    axis.legend(fontsize=8)

axes[0].set_ylabel('Monthly mean value')
fig.suptitle('Indoor-Condition Trends Across Zones', fontsize=13)
plt.tight_layout()
plt.show()

environment_summary = pd.DataFrame(environment_summary_rows)
environment_summary[['average', 'minimum', 'maximum', 'missing_percentage']] = (
    environment_summary[['average', 'minimum', 'maximum', 'missing_percentage']].round(2)
)
display(environment_summary)
```

## Key findings

- **Temperature varies moderately across zones:** Zone 1 has the lowest average at `24.75 degC`, while Zone 5 has the highest at `26.46 degC`.
- **Relative humidity is broadly similar:** zone averages range from `64.01%` in Zone 5 to `67.00%` in Zone 2.
- **Illuminance shows the clearest zone difference:** average illuminance rises from `5.91 lux` in Zone 1 to `25.26 lux` in Zone 5. Environmental columns have approximately `33%–36%` missing data, so these comparisons should be interpreted cautiously.

## Monthly temperature and AC usage

This analysis compares the average indoor temperature with total AC power across Floor 7 by month. Zone 3 is excluded because it has no temperature or AC sensors. Temperature data availability is shown because environmental sensors have substantial missing data.

```python
import numpy as np
import matplotlib.pyplot as plt

# Use zones with both temperature and AC measurements.
analysis_zones = [1, 2, 4, 5]

temperature_columns = [
    f'z{zone}_S1(degC)'
    for zone in analysis_zones
    if f'z{zone}_S1(degC)' in floor7.columns
]

ac_columns = [
    column for column in floor7.columns
    if '_AC' in column
    and column.startswith(tuple(f'z{zone}_' for zone in analysis_zones))
]

temperature_data = floor7[['Date'] + temperature_columns].copy()
temperature_data['floor_temperature_degC'] = temperature_data[temperature_columns].mean(
    axis=1,
    skipna=True
)
temperature_data['temperature_sensor_availability'] = (
    temperature_data[temperature_columns].notna().mean(axis=1) * 100
)

temperature_data['ac_usage_kW'] = floor7[ac_columns].sum(
    axis=1,
    min_count=1
)
temperature_data['ac_sensor_availability'] = (
    floor7[ac_columns].notna().mean(axis=1) * 100
)
temperature_data['month'] = temperature_data['Date'].dt.to_period('M')

monthly_temperature_ac = (
    temperature_data
    .groupby('month')
    .agg(
        average_temperature_degC=('floor_temperature_degC', 'mean'),
        average_ac_usage_kW=('ac_usage_kW', 'mean'),
        temperature_availability_pct=(
            'temperature_sensor_availability',
            'mean'
        ),
        ac_availability_pct=('ac_sensor_availability', 'mean')
    )
    .reset_index()
)

monthly_temperature_ac['month_label'] = (
    monthly_temperature_ac['month']
    .dt.to_timestamp()
    .dt.strftime('%b %Y')
)

display(monthly_temperature_ac.round(2))
```

```python
fig, axes = plt.subplots(
    2,
    1,
    figsize=(14, 9),
    sharex=True,
    gridspec_kw={'height_ratios': [3, 1]}
)

# Monthly temperature and AC comparison.
ax_temperature = axes[0]
ax_ac = ax_temperature.twinx()

ax_temperature.plot(
    monthly_temperature_ac['month_label'],
    monthly_temperature_ac['average_temperature_degC'],
    color='firebrick',
    marker='o',
    linewidth=2,
    label='Average indoor temperature'
)

ax_ac.plot(
    monthly_temperature_ac['month_label'],
    monthly_temperature_ac['average_ac_usage_kW'],
    color='steelblue',
    marker='s',
    linewidth=2,
    label='Average AC usage'
)

ax_temperature.set_ylabel('Average indoor temperature (°C)', color='firebrick')
ax_ac.set_ylabel('Average AC usage (kW)', color='steelblue')
ax_temperature.tick_params(axis='y', labelcolor='firebrick')
ax_ac.tick_params(axis='y', labelcolor='steelblue')

ax_temperature.set_title(
    'Monthly Average Indoor Temperature and AC Usage on Floor 7'
)
ax_temperature.grid(alpha=0.3)

lines_1, labels_1 = ax_temperature.get_legend_handles_labels()
lines_2, labels_2 = ax_ac.get_legend_handles_labels()
ax_temperature.legend(
    lines_1 + lines_2,
    labels_1 + labels_2,
    loc='upper left'
)

# Data availability.
axes[1].plot(
    monthly_temperature_ac['month_label'],
    monthly_temperature_ac['temperature_availability_pct'],
    color='darkorange',
    marker='o',
    label='Temperature availability'
)

axes[1].plot(
    monthly_temperature_ac['month_label'],
    monthly_temperature_ac['ac_availability_pct'],
    color='seagreen',
    marker='s',
    label='AC availability'
)

axes[1].set_ylabel('Available readings (%)')
axes[1].set_xlabel('Month')
axes[1].set_ylim(0, 105)
axes[1].grid(alpha=0.3)
axes[1].legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Use-case availability over time

This line chart tracks the percentage of available readings for each use case during September 2018 to March 2019.

```python
import matplotlib.dates as mdates

period_data = floor7.loc[
    floor7['Date'].dt.to_period('M').between(
        pd.Period('2018-09', freq='M'),
        pd.Period('2019-03', freq='M')
    )
].copy()

use_case_columns = {
    use_case: column_dictionary.loc[
        column_dictionary['use_case'] == use_case,
        'column'
    ].tolist()
    for use_case in ['AC', 'Lighting', 'Plug Load', 'Temperature', 'Humidity', 'Illuminance']
}

use_case_availability = pd.DataFrame(
    index=sorted(period_data['Date'].dt.to_period('M').unique())
)

for use_case, columns in use_case_columns.items():
    monthly_data = period_data.assign(
        month=period_data['Date'].dt.to_period('M')
    ).groupby('month')[columns]
    use_case_availability[use_case] = monthly_data.apply(
        lambda values: values.notna().mean().mean() * 100
    )

use_case_availability.index = [
    month.to_timestamp() for month in use_case_availability.index
]

plt.figure(figsize=(12, 6))
for use_case in use_case_availability.columns:
    plt.plot(
        use_case_availability.index,
        use_case_availability[use_case],
        marker='o',
        linewidth=2,
        label=use_case
    )

plt.xlabel('Month')
plt.ylabel('Available data (%)')
plt.title('Use-Case Data Availability: September 2018 to March 2019')
plt.ylim(0, 105)
plt.grid(alpha=0.3)
plt.legend(title='Use case', ncol=2)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
